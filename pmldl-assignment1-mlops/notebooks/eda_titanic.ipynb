# Titanic — краткий EDA

Разведочный анализ сырых данных (`data/raw/titanic.csv`) перед пайплайном:
пропуски, распределение целевой переменной, связь признаков с выживаемостью.
Ноутбук опционален — пайплайн его не использует.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/titanic.csv")
df.head()
```

```python
df.info()
```

```python
df.isna().sum().to_frame("missing").query("missing > 0")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
df["Survived"].value_counts().plot.bar(ax=axes[0], title="Survived")
df.groupby("Sex")["Survived"].mean().plot.bar(ax=axes[1], title="Survival rate by sex")
df.groupby("Pclass")["Survived"].mean().plot.bar(ax=axes[2], title="Survival rate by class")
plt.tight_layout()
plt.show()
```

```python
df["Fare"].plot.hist(bins=40, figsize=(7, 3), title="Fare distribution (long tail — outliers)")
plt.show()
```
